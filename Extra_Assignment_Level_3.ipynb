## Catagory the people by their age like children, adult, citizen, senior citizen...

```python
age = int(input("Enter your Age:"))
if age < 18:
    print("You are a Children")
elif (age > 18 and age < 35):
    print("You are a Adult")
elif (age > 35 and age < 59):
    print("You are a Citizen")
else:
    print("You are a Senior Citizen")
```

## Find whether given number is positive or negative

```python
num1 = int(input("Enter any Number:"))
if num1 >= 0:
    print("The Given Number is Positive")
else:
    print("The Given Number is Negative")
```

## Check whether the given number is divisible by 5

```python
num1 = int(input("Enter a Number to check it is divisible by 5:"))
if num1 % 5 == 0:
    print("The Given Number is Divisible by 5")
else:
    print("The Given Number is Not Divisible by 5")
```
