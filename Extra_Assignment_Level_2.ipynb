## Print 0 to 20 by using range ## 

```python
for i in range(0, 20):
    print(i)
```

## Print Range 10 to 20 ##

```python
for temp_i in range(10, 20):
    print(temp_i,end=" ")
```

## Concatenation of Tuples ##

```python
tuple1 = (0, 1, 2, 3)
a = tuple(tuple1)
tuple2 = ('python', 'HOPE')
b = tuple(tuple2)
print((a,b))
```

## Print Odd Numbers in the list ##

```python
list_1 = [20,10,16,19,25,1,276,188]
for i in list_1:
    if i % 2 != 0:
        print(i,"is Odd")
```

## Print Even numbers in the list ##

```python
list_2 = [20,10,16,19,25,1,276,188]
for j in list_2:
    if j % 2 == 0:
        print(j,"is Even")
```
